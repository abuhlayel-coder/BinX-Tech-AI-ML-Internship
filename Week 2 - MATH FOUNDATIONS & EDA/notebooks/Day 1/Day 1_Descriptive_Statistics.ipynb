## Central Tendency

The mean is pulled toward outliers; the median is not. Same dataset, two very different "typical values".

```python
import numpy as np

data = np.array([10, 12, 12, 13, 100])

print("mean:", np.mean(data))     # pulled up by the 100
print("median:", np.median(data)) # unaffected by the outlier
```

## Spread

Variance, standard deviation, and IQR all describe how scattered the data is - but IQR is robust to outliers while variance/std are not.

```python
print("variance:", np.var(data))
print("std dev:", np.std(data))

q1, q3 = np.percentile(data, [25, 75])
print("Q1:", q1, " Q3:", q3, " IQR:", q3 - q1)
```

## Day 1 Tasks

 the hands-on lab: In this notebook, we analyze the **Number of Votes** received by movies in the IMDb Top 1000 dataset.

The goal is to compute descriptive statistics such as the mean, median, mode, standard deviation, and interquartile range (IQR). These measures help us understand the distribution of movie popularity and determine whether the mean or median better represents a typical movie.

## Step 1 - Load a Numeric Column

We analyzed the IMDb number of votes of the top 1000 movies. The notebook computes the mean, median, mode, standard deviation, and interquartile range (IQR) to understand the distribution of movie number of votes. Finally, we compare the percentage of movies that fall below the mean and median votes to illustrate how the distribution is shaped.

```python
import pandas as pd
import numpy as np

imdb = pd.read_csv("../imdb_top_1000.csv")

votes = imdb["No_of_Votes"]

votes_np = votes.to_numpy()

print(type(votes))
print(type(votes_np))

votes.head()
```

## Step 2 - Compute Mean, Median, Mode, Standard Deviation, IQR

The statistics summarize how many IMDb users voted for each movie.

- **Mean** gives the average number of votes.
- **Median** represents the middle movie after sorting by votes.
- **Mode** is the most frequently occurring vote count.
- **Standard deviation** measures how much vote counts vary.
- **IQR** describes the spread of the middle 50% of movies and is less affected by extreme values.

```python
vote_mean = votes.mean()
vote_median = votes.median()
vote_mode = votes.mode()[0]
vote_std = votes.std()

q1, q3 = np.percentile(votes_np, [25, 75])
vote_iqr = q3 - q1

print("Mean:", round(vote_mean, 2))
print("Median:", round(vote_median, 2))
print("Mode:", vote_mode)
print("Standard Deviation:", round(vote_std, 2))
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", q3 - q1)
```

## Step 3 - Mean vs. Median: Which Is More Typical?

```python
below_mean = (votes < vote_mean).mean() * 100
below_median = (votes < vote_median).mean() * 100

print(f"{below_mean:.1f}% of movies received fewer votes than the mean.")
print(f"{below_median:.1f}% of movies received fewer votes than the median.")
```

## Justification

The **median** is a better measure of the typical number of IMDb votes than the **mean**.

The distribution of movie votes is highly skewed because a small number of blockbuster films have millions of votes, while most movies have significantly fewer. These extremely popular movies pull the mean upward, making it larger than what a typical movie receives.

The median is not strongly affected by these outliers, so it provides a more accurate representation of the typical movie in the IMDb Top 1000 dataset.
